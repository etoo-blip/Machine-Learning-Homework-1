```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("D3.csv")

print(data.head())
```

```python
#separate the variables
X1 = data["X1"].values
X2 = data["X2"].values
X3 = data["X3"].values
Y = data["Y"].values

#print the first 5 values to verify
print("X1:", X1[:5])
print("Y :", Y[:5])
```

```python
def gradient_descent(X, Y, learning_rate, iterations):

    # Initialize parameters to zero
    theta0 = 0
    theta1 = 0

    # Number of data points
    m = len(Y)

    # Store loss at every iteration
    loss_history = []

    for i in range(iterations):

        # Make predictions
        predictions = theta0 + theta1 * X

        # Calculate prediction errors
        errors = predictions - Y

        # Calculate gradients
        d_theta0 = (1 / m) * np.sum(errors)
        d_theta1 = (1 / m) * np.sum(errors * X)

        # Update parameters using gradient descent
        theta0 = theta0 - learning_rate * d_theta0
        theta1 = theta1 - learning_rate * d_theta1

        # Calculate predictions after updating theta
        new_predictions = theta0 + theta1 * X

        # Calculate cost
        cost = (1 / (2 * m)) * np.sum((new_predictions - Y) ** 2)

        # Save cost for plotting
        loss_history.append(cost)

    return theta0, theta1, loss_history
```

```python
learning_rate = 0.05
iterations = 1000
```

```python
# Train X1
theta0_X1, theta1_X1, loss_X1 = gradient_descent(
    X1, Y, learning_rate, iterations
)

# Train X2
theta0_X2, theta1_X2, loss_X2 = gradient_descent(
    X2, Y, learning_rate, iterations
)

# Train X3
theta0_X3, theta1_X3, loss_X3 = gradient_descent(
    X3, Y, learning_rate, iterations
)
```

```python
#print results
print("X1 Results")
print("theta0:", theta0_X1)
print("theta1:", theta1_X1)
print("Final Cost:", loss_X1[-1])

print()

print("X2 Results")
print("theta0:", theta0_X2)
print("theta1:", theta1_X2)
print("Final Cost:", loss_X2[-1])

print()

print("X3 Results")
print("theta0:", theta0_X3)
print("theta1:", theta1_X3)
print("Final Cost:", loss_X3[-1])
```

```python
#check change in final loss when learning rate is changed
learning_rates = [0.01, 0.05, 0.1]
iterations = 1000

variables = {
    "X1": X1,
    "X2": X2,
    "X3": X3
}

for name, X in variables.items():

    print("Results for", name)
    print("-----------------------------")

    for lr in learning_rates:

        theta0, theta1, loss = gradient_descent(
            X, Y, lr, iterations
        )

        print(
            "Learning Rate:", lr,
            "| Final Cost:", loss[-1]
        )

    print()
```

```python
#number of iterations before convergence
def convergence_iteration(loss_history, tolerance=1e-6):

    for i in range(1, len(loss_history)):

        change = abs(loss_history[i] - loss_history[i - 1])

        if change < tolerance:
            return i

    return len(loss_history)
```

```python
for name, X in variables.items():

    print("Results for", name)
    print("-----------------------------------------------")

    for lr in learning_rates:

        theta0, theta1, loss = gradient_descent(
            X, Y, lr, iterations
        )

        convergence = convergence_iteration(loss)

        print(
            "Learning Rate:", lr,
            "| Final Cost:", loss[-1],
            "| Convergence Iteration:", convergence
        )

    print()
```

```python
def gradient_descent_multiple(X1, X2, X3, Y, learning_rate, iterations):

    # Initialize all parameters to zero
    theta0 = 0
    theta1 = 0
    theta2 = 0
    theta3 = 0

    # Number of observations
    m = len(Y)

    # Store loss at every iteration
    loss_history = []

    for i in range(iterations):

        # Make predictions using all three explanatory variables
        predictions = (
            theta0
            + theta1 * X1
            + theta2 * X2
            + theta3 * X3
        )

        # Calculate errors
        errors = predictions - Y

        # Calculate gradients
        d_theta0 = (1 / m) * np.sum(errors)
        d_theta1 = (1 / m) * np.sum(errors * X1)
        d_theta2 = (1 / m) * np.sum(errors * X2)
        d_theta3 = (1 / m) * np.sum(errors * X3)

        # Update parameters
        theta0 = theta0 - learning_rate * d_theta0
        theta1 = theta1 - learning_rate * d_theta1
        theta2 = theta2 - learning_rate * d_theta2
        theta3 = theta3 - learning_rate * d_theta3

        # Calculate new predictions
        new_predictions = (
            theta0
            + theta1 * X1
            + theta2 * X2
            + theta3 * X3
        )

        # Calculate cost
        cost = (1 / (2 * m)) * np.sum((new_predictions - Y) ** 2)

        # Store cost
        loss_history.append(cost)

    return theta0, theta1, theta2, theta3, loss_history
```

```python
#training with α=0.05

learning_rate = 0.05
iterations = 1000

theta0, theta1, theta2, theta3, loss = gradient_descent_multiple(
    X1, X2, X3, Y, learning_rate, iterations
)

print("Final theta0:", theta0)
print("Final theta1:", theta1)
print("Final theta2:", theta2)
print("Final theta3:", theta3)
print("Final Cost:", loss[-1])
```

```python
#plot loss
plt.figure()

plt.plot(range(iterations), loss)

plt.xlabel("Iteration")
plt.ylabel("Cost (Loss)")
plt.title("Loss vs Iteration - Multiple Linear Regression")
plt.grid(True)

plt.show()
```

```python
#compare learning rates
learning_rates = [0.01, 0.05, 0.1]
iterations = 1000

for lr in learning_rates:

    t0, t1, t2, t3, loss_lr = gradient_descent_multiple(
        X1, X2, X3, Y, lr, iterations
    )

    convergence = convergence_iteration(loss_lr)

    print("Learning Rate:", lr)
    print("theta0:", t0)
    print("theta1:", t1)
    print("theta2:", t2)
    print("theta3:", t3)
    print("Final Cost:", loss_lr[-1])
    print("Convergence Iteration:", convergence)
    print("--------------------------------------")
```

```python
# Train final model using the selected learning rate
best_learning_rate = 0.1
iterations = 1000

theta0_best, theta1_best, theta2_best, theta3_best, loss_best = gradient_descent_multiple(
    X1, X2, X3, Y, best_learning_rate, iterations
)

print("Final Model Parameters")
print("theta0 =", theta0_best)
print("theta1 =", theta1_best)
print("theta2 =", theta2_best)
print("theta3 =", theta3_best)
print("Final Cost =", loss_best[-1])
```
